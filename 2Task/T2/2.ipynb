```python
import numpy as np
from LinearRegr import CategoricalRegression
```

```python
category = {
    1: [83, 85], 
    2: [84, 85, 85, 86, 86, 87], 
    3: [86, 87, 87, 87, 88, 88, 88, 88, 88, 89, 90], 
    4: [89, 90, 90, 91],
    5: [90, 92]
}
```

```python
model = CategoricalRegression(category=category)
coef = model.get_coefficients()
print(f'\nArray of coef:\n{np.array2string(coef, precision=2, floatmode="fixed")}')
```

```python
model = CategoricalRegression(category)

results = model.compare(alpha=0.05)

print("\nПопарные сравнения:")
print(f'{">>>":<6} {"Группы":<12} {"Разница":<10} {"t":<10} {"p-value":<10} {"α скорр.":<12} {"Значимо?":<12}')
print('-' * 70)
for r in results:
    print(f'{r["rank"]:<6} '
          f'{r["cat1"]} vs {r["cat2"]:<5} '
          f'{r["diff"]:<10.4f} '
          f'{r["t_stat"]:<10.4f} '
          f'{r["p_value"]:<10.4f} '
          f'{r["adjusted_alpha"]:<12.4f} '
          f'{"ДА" if r["significant"] else "НЕТ":<12}')
```

```python
import numpy as np

def construct_matrix(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    x = x.reshape(1,-1)
    y = y.reshape(1,-1)
    res = np.concatenate((x,y), axis = 0)
    return res

x = np.array([1,2,3])
y = np.array([2,3,4])
res = construct_matrix(x,y)
res
```
